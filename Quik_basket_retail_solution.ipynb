### Perform Calculations
|Calculations| Formula
##### |----------------------|
##### | **Total** | add up * price x quantity* for every item |
| **Discount** | 10% off when the total is more than 1000 |
##### | **Tax** | 7.5% of the amount |
| **Final figure** | amount + tax |

### return
A function can **hand back** an answer with 'return'. You can then print it, store it in a variable or give it to another function.
##### -'print()' only **shows** a value
-' return' **hand's it back** so that it can be used again|

##### Function to calculate the total

```python
# Calculate total

def calculate_total():
    total = 0       # At the initial point of calculation it starting with a total of 0
    for name in cart:  # for each item in 'cart'
        price = cart[name]['price']  # using slicing, name is the 1st key, price is 2nd Key and the value of the price
        quantity = cart[name]['quantity']
        total += price*quantity     # total = total + price* .....
    return total                # total is now a variable on it's own and can be re-used

calculate_total()
```

##### Calculate Discount
Discount is 10% of purchases up to £2000 and above

```python
# Calculating discount

def apply_discount(total):
    if total > 2000:
        return total * 0.9     # 10% off
    return total               # else return total

apply_discount(calculate_total())
```

### Calculate Tax
###### Tax of 2% -- Amount 100% + 2% = 102% = 1.02

```python
# Calculating the amount with calculated tax included

def tax_calc(total):
    return total * 1.02

tax_calc(calculate_total())
```

```python
# Calculating only the tax i.e 2%

def taxifier(total):
    return total * 0.2

taxifier(calculate_total())
```

### Final figure
##### final figure = Tax + Total


```python
# Calculating the final cost

def final_cost(total, tax):
    return total + tax

taxation = taxifier(calculate_total())
final_cost(calculate_total(), taxation)
```

# Generate Receipts

```python
def print_receipts():
    print('------------------------Receipt-----------------------------')
    print("Items : ~price~ x ~quantity~")
    print ("                         ")
    """ name is the key i,e individual purchases
    while the detail is the value i.e price and quantity 
    .items() helps to loop through each item pair  """
    
    for name, detail  in cart.items():
        line_total = detail['price'] * detail['quantity']
        print(f"{name.capitalize()} : {detail['price']} x {detail['quantity']} = {line_total}")
    print("====================================================================")

    total = calculate_total()
    total_after_discount = apply_discount(total)
    discount = total - total_after_discount
    tax = taxifier(total_after_discount)
    final_amount = tax_calc(total_after_discount)

    print(f"SUBTOTAl: {round(total,2)}")
    print(f"DISCOUNT OF 10%: {round(discount,2)}")
    print(f"AMOUNT AFTER DISCOUNT: {round(total_after_discount,2)}")
    print(f"TAX OF 2%: {round(tax,2)}")
    print(f"FINAL AMOUNT: {round(final_amount,2)}")
    print("=======================================================================")
    print(f"The Final Amount to be paid is : {round(final_amount,2)} ")
    print("                        ")
    print(f"Thank you for your patronage")

print_receipts()
```

#### Convert the above receipt into a list
###### Lists are easier to manipulate

```python
# prints are'nt saved, all prints will be added to a list with append

def line_print_receipts():
    receipt_lines = ['~~~~~~ **QUICK BASKET CHECKOUT**  ~~~~~~~~~~~~~~~~~~~'] 
    receipt_lines.append('-------------------- Receipt --------------------------')
    receipt_lines.append("Items : ~Price~ x ~Quantity~")
    receipt_lines.append("                         ")
    
    for name, detail  in cart.items():
        line_total = detail['price'] * detail['quantity']
        line = f"{name.capitalize()} : £{detail['price']} x {detail['quantity']} = £{line_total}"
        receipt_lines.append(line)
        
    line_space = "===================================================================="
    receipt_lines.append(line_space)

    total = calculate_total()
    total_after_discount = apply_discount(total)
    discount = total - total_after_discount
    tax = taxifier(total_after_discount)
    final_amount = tax_calc(total_after_discount)

    receipt_lines.append(f"SUBTOTAl: £{round(total,2)}")
    receipt_lines.append(f"DISCOUNT OF 10%: £{round(discount,2)}")
    receipt_lines.append(f"AMOUNT AFTER DISCOUNT: £{round(total_after_discount,2)}")
    receipt_lines.append(f"TAX OF 2%: £{round(tax,2)}")
    receipt_lines.append(f"FINAL AMOUNT: £{round(final_amount,2)}")
    receipt_lines.append("=======================================================================")
    receipt_lines.append(f"The Final Amount to be paid is : £{round(final_amount,2)} ")
    receipt_lines.append("                        ")
    receipt_lines.append(f"Thank you for your patronage")
    return receipt_lines
```

```python
# To remove the (,) effect of the list, we use a loop to print each line

lines = line_print_receipts()
for i in lines:
    print(i)
```

```python
# Function to save the receipt as a PNG

def save_receipt(lines, filename = 'receipt.png'):
    filename_prompt = input('Enter the transaction name: ')
    if filename_prompt != '':                                # If the filename is not empty
        filename = filename_prompt + '.png'

    # picture size
    width = 400
    line_height = 25
    height = line_height * (len(lines) + 2)                  # len(lines) + 2 --- length of the lines list + 2 extra spaces

    # create image
    image = Image.new('RGB', (width, height), 'white')
    draw = ImageDraw.Draw(image)

    try:                                                     # Try the next line
        font = ImageFont.truetype('arial.ttf', 16)
    except OSError:                                          # If the above Arial font is not on this computer, use the default font
        font = ImageFont.load_default(size = 16)

    # draw each line
    y = 10
    for line in lines:
        draw.text((10, y), line, fill = 'black', font = font)
        y += line_height
    image.save(filename)
    print(f"Receipt saved as {filename}")
    
    
```

```python
save_receipt(lines)
```
